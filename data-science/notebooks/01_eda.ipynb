Analyzes transaction volume, fraud prevalence (<1% on Sparkov dataset), fraud rate by merchant category and hour of day, transaction amount distributions, and train/test date ranges.

```python
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

REPO_ROOT = Path.cwd().resolve().parents[1]
sys.path.insert(0, str(REPO_ROOT))
sys.path.insert(0, str(REPO_ROOT / 'ml-service'))

from training.train import load_train_test_splits

train_df, test_df = load_train_test_splits(seed=42)
print(f'Train rows: {len(train_df):,} | range: {train_df["tx_time"].min()} -> {train_df["tx_time"].max()}')
print(f'Test  rows: {len(test_df):,}  | range: {test_df["tx_time"].min()} -> {test_df["tx_time"].max()}')
print(f'Train fraud rate: {train_df["is_fraud"].mean()*100:.3f}% | Test fraud rate: {test_df["is_fraud"].mean()*100:.3f}%')
```

```python
# Fraud rate by category and hour of day
train_df['hour'] = pd.to_datetime(train_df['tx_time'], utc=True).dt.hour
cat_summary = train_df.groupby('category')['is_fraud'].agg(['count', 'mean']).sort_values('mean', ascending=False)
hour_summary = train_df.groupby('hour')['is_fraud'].mean()
display(cat_summary)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
cat_summary['mean'].plot(kind='bar', ax=axes[0], title='Fraud Rate by Category')
hour_summary.plot(kind='line', marker='o', ax=axes[1], title='Fraud Rate by UTC Hour')
plt.tight_layout()
```
